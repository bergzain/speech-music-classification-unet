```python
import pandas as pd
file_path = '/Users/zainhazzouri/projects/Bachelor_Thesis/results/runs.xlsx'
# Load the CSV file into a DataFrame
data = pd.read_excel(file_path)

# Display the first 5 rows
print(data.head())
```

```python
is_unique = data['Name'].is_unique
if is_unique:
    print("The Name column contains unique values.")
else:
    print("The 'Name column has duplicate values.")
duplicates = data[data.duplicated(subset=["Name"], keep=False)]
if not duplicates.empty:
    print("Duplicate Name:")
    print(duplicates['Name'].unique())
```

```python
import os
from openpyxl import load_workbook
from openpyxl.styles import PatternFill
import pandas as pd


# Find the indices of all rows with the best validation accuracy
max_val_accuracy = data['best_val_accuracy'].max()
best_val_accuracy_indices = data.index[data['best_val_accuracy'] == max_val_accuracy].tolist()

# Find the indices of all rows with the least validation loss
min_val_loss = data['val_loss'].min()
least_val_loss_indices = data.index[data['val_loss'] == min_val_loss].tolist()

# Load the workbook and select the active sheet
workbook = load_workbook(file_path)
sheet = workbook.active

# Define highlight styles
accuracy_fill = PatternFill(start_color="FFFF00", end_color="FFFF00", fill_type="solid")  # Yellow fill for accuracy
loss_fill = PatternFill(start_color="00FF00", end_color="00FF00", fill_type="solid")  # Green fill for loss

# Highlight all rows with the best validation accuracy
for row_idx in best_val_accuracy_indices:
    for cell in sheet[row_idx + 2]:  # Offset for header row
        cell.fill = accuracy_fill

# Highlight all rows with the least validation loss
for row_idx in least_val_loss_indices:
    for cell in sheet[row_idx + 2]:  # Offset for header row
        cell.fill = loss_fill

# Construct output file path
dir_name, base_name = os.path.split(file_path)
output_file = os.path.join(dir_name, f"highlighted_{base_name}")

# Save the updated Excel file
workbook.save(output_file)
print(f"Rows with the best validation accuracy and least loss highlighted successfully in: {output_file}")
```
